```python
import flet as ft
import time
import random

def main(page: ft.Page):
    page.title = "Sala 4 - Nanotecnología y Chips Cuánticos"
    page.bgcolor = "#E8E8E8"
    page.padding = 0
    page.spacing = 0

    nivel_actual = 0

    # 4 AUDIOS desde GitHub
    audio_urls = [
        "https://raw.githubusercontent.com/Prof-Luis1986/Proyctos_3roB/Prof-Luis1986-patch-1/Audios_Videos/Video1.mp3",
        "https://raw.githubusercontent.com/Prof-Luis1986/Proyctos_3roB/Prof-Luis1986-patch-1/Audios_Videos/Video2.mp3",
        "https://raw.githubusercontent.com/Prof-Luis1986/Proyctos_3roB/Prof-Luis1986-patch-1/Audios_Videos/video3.mp3",
        "https://raw.githubusercontent.com/Prof-Luis1986/Proyctos_3roB/Prof-Luis1986-patch-1/Audios_Videos/video4.mp3",
    ]

    # Definición de niveles con colores del proyecto
    niveles = [
        {
            "nombre": "🔲 NIVEL MACRO",
            "subtitulo": "Placa de Circuito",
            "imagen": "https://images.unsplash.com/photo-1518770660439-4636190af475?w=800&q=80",
            "descripcion": "Todo comenzó con las placas de circuito. Componentes visibles, cables, resistencias... La tecnología que podíamos tocar y ver.",
            "color": "#8B3A4A",
            "icono": "🔲",
            "emoji_grande": "💻",
            "dato_curioso": "💡 Las primeras computadoras ocupaban habitaciones enteras"
        },
        {
            "nombre": "🔬 NIVEL MICRO",
            "subtitulo": "La Revolución del Silicio",
            "imagen": "https://images.unsplash.com/photo-1555617981-dac3880eac6e?w=800&q=80",
            "descripcion": "Luego llegó la miniaturización. Millones de transistores en un chip del tamaño de una uña. La Ley de Moore predijo que duplicaríamos la potencia cada 2 años.",
            "color": "#A64D5A",
            "icono": "🔬",
            "emoji_grande": "🔍",
            "dato_curioso": "💡 Un chip moderno tiene más de 50 mil millones de transistores"
        },
        {
            "nombre": "⚛️ NIVEL NANO",
            "subtitulo": "El Límite de lo Clásico",
            "imagen": "https://images.unsplash.com/photo-1635070041078-e363dbe005cb?w=800&q=80",
            "descripcion": "Pero llegamos al límite. A escala nanométrica, los electrones hacen 'túnel cuántico' atravesando barreras. Las leyes clásicas ya no funcionan.",
            "color": "#6B2A3A",
            "icono": "⚛️",
            "emoji_grande": "✨",
            "dato_curioso": "💡 A 5 nanómetros, solo caben 20 átomos de ancho"
        },
        {
            "nombre": "🌀 NIVEL CUÁNTICO",
            "subtitulo": "Microsoft y el Futuro Cuántico",
            "imagen": "https://images.unsplash.com/photo-1635070041409-e63e783ce3b0?w=800&q=80",
            "descripcion": "Microsoft Azure Quantum lidera la revolución. Sus qubits topológicos prometen ser más estables. Un qubit puede procesar múltiples estados simultáneamente. ¡El futuro ya está aquí!",
            "color": "#C17A89",
            "icono": "🌀",
            "emoji_grande": "🚀",
            "dato_curioso": "💡 Microsoft invierte miles de millones en computación cuántica"
        }
    ]

    # AUDIO PLAYER - Sin deprecation warning
    audio_player = ft.Audio(
        src=audio_urls[0],
        autoplay=True,
        volume=1.0,
    )
    page.overlay.append(audio_player)

    # HEADER
    header = ft.Container(
        content=ft.Column([
            ft.Row([
                ft.Icon(ft.Icons.MUSEUM, color="#FFFFFF", size=40),
                ft.Text(
                    "MUSEO DE LA INFORMÁTICA - CETis 50",
                    size=20,
                    color="#FFFFFF",
                    weight=ft.FontWeight.BOLD
                )
            ], alignment=ft.MainAxisAlignment.CENTER),
            ft.Text(
                "Sala 4: Nanotecnología y Chips Cuánticos",
                size=16,
                color="#F5E6D3",
                text_align=ft.TextAlign.CENTER,
                italic=True
            )
        ], horizontal_alignment=ft.CrossAxisAlignment.CENTER, spacing=5),
        bgcolor="#8B3A4A",
        padding=20,
        border_radius=0
    )

    titulo = ft.Container(
        content=ft.Text(
            "🔬 SALA 4: NANOTECNOLOGÍA Y CHIPS CUÁNTICOS",
            size=40,
            weight=ft.FontWeight.BOLD,
            color="#8B3A4A",
            text_align=ft.TextAlign.CENTER,
        ),
        animate=600,
        padding=ft.padding.only(top=20, bottom=10)
    )

    subtitulo = ft.Text(
        "🎮 ¡Presiona las teclas para hacer ZOOM en el chip!",
        size=20,
        color="#6B2A3A",
        text_align=ft.TextAlign.CENTER,
        weight=ft.FontWeight.BOLD
    )

    emoji_grande = ft.Text(
        niveles[0]["emoji_grande"],
        size=100,
        text_align=ft.TextAlign.CENTER,
        animate_scale=800,
        animate_rotation=600
    )

    nivel_texto = ft.Container(
        content=ft.Text(
            niveles[0]["nombre"],
            size=36,
            weight=ft.FontWeight.BOLD,
            color="#FFFFFF",
            text_align=ft.TextAlign.CENTER,
        ),
        bgcolor=niveles[0]["color"],
        padding=15,
        border_radius=15,
        shadow=ft.BoxShadow(
            spread_radius=5,
            blur_radius=20,
            color=niveles[0]["color"] + "66",
        ),
        animate=400
    )

    subtitulo_nivel = ft.Text(
        niveles[0]["subtitulo"],
        size=24,
        color="#6B2A3A",
        text_align=ft.TextAlign.CENTER,
        italic=True,
        weight=ft.FontWeight.BOLD
    )

    # Indicador de audio con animación
    audio_icon = ft.Icon(
        ft.Icons.VOLUME_UP_ROUNDED,
        color="#8B3A4A",
        size=40,
        animate_scale=1000
    )

    audio_status = ft.Container(
        content=ft.Row([
            audio_icon,
            ft.Text("🎵 Audio explicativo en reproducción", size=18, color="#8B3A4A", weight=ft.FontWeight.BOLD)
        ], alignment=ft.MainAxisAlignment.CENTER),
        bgcolor="#F5E6D3",
        padding=12,
        border_radius=10,
        border=ft.border.all(2, "#8B3A4A"),
        animate_opacity=300
    )

    # Animación del icono de audio
    def animar_audio_icon():
        while True:
            audio_icon.scale = 1.2
            page.update()
            time.sleep(0.5)
            audio_icon.scale = 1.0
            page.update()
            time.sleep(0.5)

    import threading
    audio_thread = threading.Thread(target=animar_audio_icon, daemon=True)
    audio_thread.start()

    imagen_nivel = ft.Container(
        content=ft.Image(
            src=niveles[0]["imagen"],
            width=700,
            height=450,
            fit=ft.ImageFit.COVER,
        ),
        border_radius=25,
        border=ft.border.all(3, "#8B3A4A"),
        shadow=ft.BoxShadow(
            spread_radius=10,
            blur_radius=30,
            color="#8B3A4A44",
        ),
        animate_scale=500,
        animate=500
    )

    descripcion_texto = ft.Container(
        content=ft.Text(
            niveles[0]["descripcion"],
            size=20,
            color="#333333",
            text_align=ft.TextAlign.CENTER,
            weight=ft.FontWeight.W_500
        ),
        bgcolor="#FFFFFF",
        padding=20,
        border_radius=15,
        width=850,
        border=ft.border.all(2, "#8B3A4A"),
        animate_opacity=300
    )

    dato_curioso = ft.Container(
        content=ft.Text(
            niveles[0]["dato_curioso"],
            size=18,
            color="#6B2A3A",
            text_align=ft.TextAlign.CENTER,
            weight=ft.FontWeight.BOLD
        ),
        bgcolor="#F5E6D3",
        padding=15,
        border_radius=10,
        width=850,
        border=ft.border.all(2, "#8B3A4A"),
        animate_opacity=300
    )

    indicadores = ft.Row(
        controls=[],
        alignment=ft.MainAxisAlignment.CENTER,
        spacing=30
    )

    def crear_indicadores():
        indicadores.controls.clear()
        for i in range(4):
            color = niveles[i]["color"]
            activo = i == nivel_actual
            completado = i < nivel_actual

            container = ft.Container(
                content=ft.Column([
                    ft.Text(
                        niveles[i]["icono"],
                        size=40 if activo else 30,
                        text_align=ft.TextAlign.CENTER
                    ),
                    ft.Text(
                        f"Nivel {i+1}",
                        size=12,
                        color="#FFFFFF" if (activo or completado) else "#999999",
                        text_align=ft.TextAlign.CENTER,
                        weight=ft.FontWeight.BOLD
                    )
                ], horizontal_alignment=ft.CrossAxisAlignment.CENTER, spacing=5),
                width=90,
                height=90,
                bgcolor=color if activo else ("#FFFFFF" if not completado else color + "44"),
                border=ft.border.all(4 if activo else 2, color if (activo or completado) else "#CCCCCC"),
                border_radius=45,
                alignment=ft.alignment.center,
                shadow=ft.BoxShadow(
                    spread_radius=5 if activo else 0,
                    blur_radius=20 if activo else 5,
                    color=color + "88" if activo else "#00000022",
                ),
                animate=400,
                animate_scale=300
            )

            if activo:
                container.scale = 1.1

            indicadores.controls.append(container)

    crear_indicadores()

    particulas = ft.Row(
        controls=[
            ft.Text("✨", size=20, opacity=0.6),
            ft.Text("⚡", size=20, opacity=0.4),
            ft.Text("💫", size=20, opacity=0.5),
            ft.Text("🌟", size=20, opacity=0.7),
            ft.Text("✨", size=20, opacity=0.3),
        ],
        alignment=ft.MainAxisAlignment.CENTER,
        spacing=50
    )

    bit_texto = ft.Text("1", size=70, weight=ft.FontWeight.BOLD, color="#8B3A4A")

    bit_clasico = ft.Container(
        content=ft.Column([
            ft.Text("⚡ BIT CLÁSICO", size=26, weight=ft.FontWeight.BOLD, color="#8B3A4A"),
            ft.Container(
                content=bit_texto,
                width=150,
                height=150,
                bgcolor="#F5E6D3",
                border=ft.border.all(4, "#8B3A4A"),
                border_radius=75,
                alignment=ft.alignment.center,
                shadow=ft.BoxShadow(
                    spread_radius=5,
                    blur_radius=20,
                    color="#8B3A4A66",
                ),
            ),
            ft.Text("Solo puede ser 0 ó 1", size=18, color="#333333", text_align=ft.TextAlign.CENTER),
            ft.Text("(Uno a la vez)", size=14, color="#666666", text_align=ft.TextAlign.CENTER, italic=True)
        ], horizontal_alignment=ft.CrossAxisAlignment.CENTER, spacing=15),
        visible=False,
        animate_opacity=300,
        animate_scale=500
    )

    qubit_texto = ft.Text("0+1", size=70, weight=ft.FontWeight.BOLD, color="#6B2A3A")

    qubit_container = ft.Container(
        content=qubit_texto,
        width=150,
        height=150,
        bgcolor="#F5E6D3",
        border=ft.border.all(4, "#6B2A3A"),
        border_radius=75,
        alignment=ft.alignment.center,
        shadow=ft.BoxShadow(
            spread_radius=5,
            blur_radius=20,
            color="#6B2A3A66",
        ),
        animate=300,
        animate_rotation=500
    )

    qubit_visual = ft.Container(
        content=ft.Column([
            ft.Text("🌀 QUBIT CUÁNTICO", size=26, weight=ft.FontWeight.BOLD, color="#6B2A3A"),
            qubit_container,
            ft.Text("¡Puede ser 0 Y 1 al mismo tiempo!", size=18, color="#333333", text_align=ft.TextAlign.CENTER),
            ft.Text("(Superposición cuántica)", size=14, color="#666666", text_align=ft.TextAlign.CENTER, italic=True)
        ], horizontal_alignment=ft.CrossAxisAlignment.CENTER, spacing=15),
        visible=False,
        animate_opacity=300,
        animate_scale=500
    )

    comparacion_quantum = ft.Container(
        content=ft.Row(
            controls=[bit_clasico, qubit_visual],
            alignment=ft.MainAxisAlignment.CENTER,
            spacing=120
        ),
        bgcolor="#FFFFFF",
        padding=30,
        border_radius=20,
        border=ft.border.all(3, "#8B3A4A"),
        shadow=ft.BoxShadow(
            spread_radius=5,
            blur_radius=15,
            color="#8B3A4A33",
        )
    )

    instrucciones = ft.Container(
        content=ft.Column([
            ft.Text("🎮 CONTROLES MAKEY MAKEY", size=24, weight=ft.FontWeight.BOLD, color="#FFFFFF"),
            ft.Divider(height=10, color="#FFFFFF44"),
            ft.Row([
                ft.Container(
                    content=ft.Text("↓", size=30, color="#FFFFFF"),
                    bgcolor="#6B2A3A",
                    width=50,
                    height=50,
                    border_radius=10,
                    alignment=ft.alignment.center,
                    border=ft.border.all(2, "#FFFFFF")
                ),
                ft.Text("FLECHA ABAJO: Profundizar al siguiente nivel", size=18, color="#FFFFFF")
            ], spacing=15),
            ft.Row([
                ft.Container(
                    content=ft.Text("↑", size=30, color="#FFFFFF"),
                    bgcolor="#6B2A3A",
                    width=50,
                    height=50,
                    border_radius=10,
                    alignment=ft.alignment.center,
                    border=ft.border.all(2, "#FFFFFF")
                ),
                ft.Text("FLECHA ARRIBA: Regresar al nivel anterior", size=18, color="#FFFFFF")
            ], spacing=15),
            ft.Row([
                ft.Container(
                    content=ft.Text("⎵", size=30, color="#FFFFFF"),
                    bgcolor="#6B2A3A",
                    width=50,
                    height=50,
                    border_radius=10,
                    alignment=ft.alignment.center,
                    border=ft.border.all(2, "#FFFFFF")
                ),
                ft.Text("ESPACIO: Reiniciar al nivel macro", size=18, color="#FFFFFF")
            ], spacing=15),
        ], horizontal_alignment=ft.CrossAxisAlignment.CENTER, spacing=10),
        bgcolor="#8B3A4A",
        padding=25,
        border_radius=15,
        shadow=ft.BoxShadow(
            spread_radius=5,
            blur_radius=15,
            color="#8B3A4A44",
        )
    )

    zoom_counter = ft.Text(
        "🔍 ZOOM: 1x",
        size=28,
        weight=ft.FontWeight.BOLD,
        color="#8B3A4A",
        text_align=ft.TextAlign.CENTER
    )

    # Contenedor principal de contenido
    contenido_principal = ft.Column(
        controls=[
            titulo,
            subtitulo,
            ft.Container(height=10),
            audio_status,
            ft.Container(height=10),
            zoom_counter,
            ft.Container(height=10),
            emoji_grande,
            ft.Container(height=10),
            nivel_texto,
            subtitulo_nivel,
            ft.Container(height=20),
            imagen_nivel,
            ft.Container(height=20),
            particulas,
            ft.Container(height=10),
            descripcion_texto,
            ft.Container(height=10),
            dato_curioso,
            ft.Container(height=25),
            indicadores,
            ft.Container(height=25),
            comparacion_quantum,
            ft.Container(height=25),
            instrucciones,
            ft.Container(height=20),
        ],
        horizontal_alignment=ft.CrossAxisAlignment.CENTER,
        scroll=ft.ScrollMode.AUTO
    )

    def reproducir_audio_nivel(nivel):
        try:
            audio_player.src = audio_urls[nivel]
            audio_player.update()
            print(f"🎵 Reproduciendo audio {nivel + 1}: {audio_urls[nivel]}")
        except Exception as e:
            print(f"❌ Error reproduciendo audio: {e}")

    def animar_bit():
        for _ in range(6):
            bit_texto.value = "0" if bit_texto.value == "1" else "1"
            bit_texto.color = "#C17A89" if bit_texto.value == "0" else "#8B3A4A"
            page.update()
            time.sleep(0.3)

    def animar_qubit():
        if nivel_actual == 3:
            estados = ["0", "1", "0+1", "∞", "⚛️", "🌀", "0&1", "???"]
            colores = ["#8B3A4A", "#A64D5A", "#6B2A3A", "#C17A89", "#8B3A4A", "#A64D5A", "#6B2A3A", "#C17A89"]

            for i in range(12):
                estado = random.choice(estados)
                color = random.choice(colores)

                qubit_texto.value = estado
                qubit_texto.color = color
                qubit_container.border = ft.border.all(4, color)
                qubit_container.bgcolor = "#F5E6D3"
                qubit_container.shadow = ft.BoxShadow(
                    spread_radius=8,
                    blur_radius=25,
                    color=color + "88",
                )
                qubit_container.rotate = (i * 0.2) % (2 * 3.14159)
                page.update()
                time.sleep(0.25)

    def actualizar_nivel():
        nonlocal nivel_actual
        nivel = nivel_actual

        reproducir_audio_nivel(nivel)

        zoom_levels = ["1x", "1,000x", "1,000,000x", "∞"]
        zoom_counter.value = f"🔍 ZOOM: {zoom_levels[nivel]}"

        emoji_grande.value = niveles[nivel]["emoji_grande"]
        emoji_grande.scale = 1.5
        emoji_grande.rotate = 0.3
        page.update()
        time.sleep(0.1)
        emoji_grande.scale = 1.0
        emoji_grande.rotate = 0

        nivel_texto.content.value = niveles[nivel]["nombre"]
        nivel_texto.bgcolor = niveles[nivel]["color"]
        nivel_texto.shadow = ft.BoxShadow(
            spread_radius=5,
            blur_radius=20,
            color=niveles[nivel]["color"] + "66",
        )

        subtitulo_nivel.value = niveles[nivel]["subtitulo"]

        imagen_nivel.content.src = niveles[nivel]["imagen"]
        imagen_nivel.scale = 1.2
        imagen_nivel.border = ft.border.all(3, niveles[nivel]["color"])
        imagen_nivel.shadow = ft.BoxShadow(
            spread_radius=10,
            blur_radius=30,
            color=niveles[nivel]["color"] + "44",
        )
        page.update()
        time.sleep(0.2)
        imagen_nivel.scale = 1.0

        descripcion_texto.content.value = niveles[nivel]["descripcion"]
        descripcion_texto.border = ft.border.all(2, niveles[nivel]["color"])

        dato_curioso.content.value = niveles[nivel]["dato_curioso"]

        crear_indicadores()

        bit_clasico.visible = (nivel == 3)
        qubit_visual.visible = (nivel == 3)

        page.update()

        if nivel == 3:
            time.sleep(0.3)
            animar_bit()
            animar_qubit()

    def on_keyboard(e: ft.KeyboardEvent):
        nonlocal nivel_actual

        if e.key == "Arrow Down" or e.key == "S" or e.key == "s":
            if nivel_actual < 3:
                nivel_actual += 1
                actualizar_nivel()

        elif e.key == "Arrow Up" or e.key == "W" or e.key == "w":
            if nivel_actual > 0:
                nivel_actual -= 1
                actualizar_nivel()

        elif e.key == " " or e.key == "Space":
            nivel_actual = 0
            actualizar_nivel()

    page.on_keyboard_event = on_keyboard

    page.add(
        header,
        ft.Container(
            content=contenido_principal,
            padding=20,
            expand=True
        )
    )

    print("✅ Aplicación iniciada con audios desde GitHub")
    print(f"🎵 Audio 1 reproduciéndose: {audio_urls[0]}")

ft.app(target=main)
```
