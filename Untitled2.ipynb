```python
import flet as ft
import pygame
import requests
import tempfile
import os

# ============================================================
# 🎨 PALETA INSTITUCIONAL CETIS
# ============================================================
COLOR_GUINDA = "#7A1C1C"
COLOR_BLANCO = "#FFFFFF"
COLOR_FONDO = "#F9F9F9"
COLOR_TEXTO = "#2C2C2C"
COLOR_SUBTEXTO = "#555555"
COLOR_TARJETA = "#EDEDED"

# ============================================================
# 🎧 Inicializar pygame mixer para reproducir audios
# ============================================================
pygame.mixer.init()

def reproducir_audio(url):
    """Descarga temporalmente y reproduce el audio"""
    detener_audio()
    try:
        response = requests.get(url)
        response.raise_for_status()
        with tempfile.NamedTemporaryFile(delete=False, suffix=".mp3") as temp_audio:
            temp_audio.write(response.content)
            temp_path = temp_audio.name
        pygame.mixer.music.load(temp_path)
        pygame.mixer.music.play()
    except Exception as e:
        print(f"Error reproduciendo: {e}")

def detener_audio():
    """Detiene cualquier audio en curso"""
    try:
        if pygame.mixer.music.get_busy():
            pygame.mixer.music.stop()
    except:
        pass

# ============================================================
# 💾 DATOS DE LOS MEDIOS (imágenes/audios del proyecto)
# ============================================================
TIMELINE_DATA = [
    {
        "id": "A", "nombre": "Tarjeta perforada", "decada": "1950s",
        "capacidad": "80 bytes", "capacidad_bytes": 80,
        "curiosidad": "Una de las primeras formas de almacenamiento. Se usaban para guardar programas y datos.",
        "imagen": "https://raw.githubusercontent.com/Prof-Luis1986/Proyctos_3roB/refs/heads/main/Tarjeta_perforada.jpg",
        "sonido": "https://raw.githubusercontent.com/Prof-Luis1986/Proyctos_3roB/refs/heads/main/tarjeta%20perforada.mp3"
    },
    {
        "id": "S", "nombre": "Disquete (Floppy)", "decada": "1980s",
        "capacidad": "1.44 MB", "capacidad_bytes": 1_440_000,
        "curiosidad": "El famoso ícono de 'Guardar' viene de este dispositivo.",
        "imagen": "https://raw.githubusercontent.com/Prof-Luis1986/Proyctos_3roB/refs/heads/main/floppy.jpg",
        "sonido": "https://raw.githubusercontent.com/Prof-Luis1986/Proyctos_3roB/refs/heads/main/floppy.mp3"
    },
    {
        "id": "D", "nombre": "CD / DVD", "decada": "1990s",
        "capacidad": "700 MB / 4.7 GB", "capacidad_bytes": 700_000_000,
        "curiosidad": "Permitieron almacenar música, películas y software de forma masiva.",
        "imagen": "https://raw.githubusercontent.com/Prof-Luis1986/Proyctos_3roB/refs/heads/main/cd.jpg",
        "sonido": "https://raw.githubusercontent.com/Prof-Luis1986/Proyctos_3roB/refs/heads/main/CD.mp3"
    },
    {
        "id": "F", "nombre": "USB", "decada": "2000s",
        "capacidad": "4 GB - 128 GB", "capacidad_bytes": 64_000_000_000,
        "curiosidad": "Popularizó el almacenamiento portátil y rápido.",
        "imagen": "https://raw.githubusercontent.com/Prof-Luis1986/Proyctos_3roB/refs/heads/main/usb.jpg",
        "sonido": "https://raw.githubusercontent.com/Prof-Luis1986/Proyctos_3roB/refs/heads/main/USB.mp3"
    },
    {
        "id": "G", "nombre": "SSD", "decada": "2010s",
        "capacidad": "256 GB - 4 TB", "capacidad_bytes": 1_000_000_000_000,
        "curiosidad": "Sin partes móviles, ofrece una velocidad increíble comparada con los discos duros.",
        "imagen": "https://raw.githubusercontent.com/Prof-Luis1986/Proyctos_3roB/refs/heads/main/ssd.jpg",
        "sonido": "https://raw.githubusercontent.com/Prof-Luis1986/Proyctos_3roB/refs/heads/main/SSD.mp3"
    },
    {
        "id": "H", "nombre": "Nube", "decada": "2020s",
        "capacidad": "Ilimitada", "capacidad_bytes": 10_000_000_000_000,
        "curiosidad": "Permite guardar datos en servidores remotos accesibles desde cualquier lugar.",
        "imagen": "https://raw.githubusercontent.com/Prof-Luis1986/Proyctos_3roB/refs/heads/main/Nube.jpeg",
        "sonido": "https://raw.githubusercontent.com/Prof-Luis1986/Proyctos_3roB/refs/heads/main/NUBE.mp3"
    }
]

CAPACIDAD_ACTUAL = 10_000_000_000_000

# ============================================================
# 🧠 APP PRINCIPAL
# ============================================================
def main(page: ft.Page):
    page.title = "Sala 2 — Evolución del almacenamiento | CETIS 50"
    page.theme_mode = ft.ThemeMode.LIGHT
    page.bgcolor = COLOR_FONDO
    page.padding = 20
    page.fonts = {
        "Roboto": "https://fonts.gstatic.com/s/roboto/v30/KFOmCnqEu92Fr1Me5Q.woff2"
    }
    page.theme = ft.Theme(font_family="Roboto")

    shortcuts = {m["id"]: m for m in TIMELINE_DATA}

    # =============== VISTA INICIO ==================
    def build_home():
        detener_audio()
        return ft.View(
            "/",
            [
                ft.Container(
                    bgcolor=COLOR_GUINDA,
                    border_radius=10,
                    padding=20,
                    content=ft.Column([
                        ft.Text("Proyecto Sala 2", size=24, color=COLOR_BLANCO, weight="bold"),
                        ft.Text("“Evolución de los Medios de Almacenamiento”", size=20, color=COLOR_BLANCO)
                    ], horizontal_alignment="center")
                ),
                ft.Container(height=30),
                ft.Text("Selecciona un medio o usa el Makey Makey (A-H)", color=COLOR_TEXTO, size=18, text_align="center"),
                ft.Container(height=20),
                ft.ElevatedButton(
                    "🚀 Entrar a la Línea del Tiempo",
                    bgcolor=COLOR_GUINDA,
                    color=COLOR_BLANCO,
                    height=60,
                    width=350,
                    style=ft.ButtonStyle(shape=ft.RoundedRectangleBorder(radius=10),
                                         text_style=ft.TextStyle(size=22, weight="bold")),
                    on_click=lambda e: page.go("/timeline")
                )
            ],
            horizontal_alignment="center",
            vertical_alignment="center"
        )

    # =============== VISTA LÍNEA DE TIEMPO ==================
    def build_timeline():
        detener_audio()
        cards = []
        for m in TIMELINE_DATA:
            cards.append(
                ft.Container(
                    bgcolor=COLOR_TARJETA,
                    border_radius=15,
                    padding=10,
                    width=170,
                    on_click=lambda e, mid=m["id"]: page.go(f"/detail/{mid}"),
                    content=ft.Column([
                        ft.Image(src=m["imagen"], width=150, height=120, fit=ft.ImageFit.COVER, border_radius=10),
                        ft.Text(m["nombre"], color=COLOR_GUINDA, weight="bold", size=16, text_align="center"),
                        ft.Text(m["decada"], color=COLOR_SUBTEXTO, size=14, text_align="center")
                    ], horizontal_alignment="center", spacing=4)
                )
            )

        return ft.View(
            "/timeline",
            [
                ft.Text("📅 Línea del tiempo", size=28, weight="bold", color=COLOR_GUINDA, text_align="center"),
                ft.Container(height=20),
                ft.Row(cards, wrap=True, alignment="center", spacing=15),
                ft.Container(height=30),
                ft.ElevatedButton("🏠 Volver al inicio", on_click=lambda e: page.go("/"),
                                  bgcolor=COLOR_GUINDA, color=COLOR_BLANCO, height=50, width=250)
            ],
            horizontal_alignment="center",
            vertical_alignment="center"
        )

    # =============== VISTA DETALLE ==================
    def build_detail(medio_id):
        medio = shortcuts.get(medio_id)
        if not medio:
            return ft.View("/", [ft.Text("⚠️ No encontrado", color=COLOR_GUINDA)])

        # 🔊 Reproduce el audio automáticamente al entrar
        reproducir_audio(medio["sonido"])

        cap_bytes = medio["capacidad_bytes"]
        veces_cabe = CAPACIDAD_ACTUAL / cap_bytes

        def format_bytes(val):
            if val >= 1_000_000_000_000:
                return f"{val/1_000_000_000_000:.1f} TB"
            elif val >= 1_000_000_000:
                return f"{val/1_000_000_000:.1f} GB"
            elif val >= 1_000_000:
                return f"{val/1_000_000:.1f} MB"
            elif val >= 1_000:
                return f"{val/1_000:.1f} KB"
            else:
                return f"{val} bytes"

        return ft.View(
            f"/detail/{medio_id}",
            [
                ft.Container(
                    bgcolor=COLOR_TARJETA,
                    padding=20,
                    border_radius=15,
                    content=ft.Column([
                        ft.Text(medio["nombre"], size=34, weight="bold", color=COLOR_GUINDA, text_align="center"),
                        ft.Text(f"({medio['decada']})", size=22, color=COLOR_SUBTEXTO, text_align="center"),
                        ft.Container(height=15),
                        ft.Image(src=medio["imagen"], width=450, height=300, fit=ft.ImageFit.CONTAIN),
                        ft.Text(medio["curiosidad"], size=20, color=COLOR_TEXTO, text_align="center"),
                        ft.Text(f"💾 Capacidad: {medio['capacidad']}", size=18, color=COLOR_GUINDA),
                        ft.Text(f"🌩️ En la nube cabrían {veces_cabe:,.0f} de estos medios.", size=16, color=COLOR_SUBTEXTO, text_align="center"),
                        ft.Container(height=20),
                        ft.ElevatedButton(
                            "⬅️ Volver",
                            on_click=lambda e: (detener_audio(), page.go("/timeline")),
                            bgcolor=COLOR_GUINDA, color=COLOR_BLANCO, height=50, width=200
                        )
                    ], horizontal_alignment="center", alignment="center", spacing=10)
                )
            ],
            horizontal_alignment="center",
            vertical_alignment="center"
        )

    # =============== RUTAS ==================
    def route_change(e):
        page.views.clear()
        detener_audio()  # detiene audio al cambiar de pantalla
        if page.route == "/":
            page.views.append(build_home())
        elif page.route == "/timeline":
            page.views.append(build_timeline())
        elif page.route.startswith("/detail/"):
            medio_id = page.route.split("/")[-1]
            page.views.append(build_detail(medio_id))
        page.update()

    # Teclado (Makey Makey)
    def on_keyboard(e: ft.KeyboardEvent):
        key = e.key.upper()
        if key in shortcuts:
            page.go(f"/detail/{key}")

    page.on_route_change = route_change
    page.on_keyboard_event = on_keyboard
    page.go("/")

# ============================================================
# 🚀 EJECUTAR
# ============================================================
ft.app(target=main)
```
