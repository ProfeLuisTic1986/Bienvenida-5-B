🧱 4. Escribe el código paso a paso
* Paso 1️⃣: Importar Flet

    Abre main.py y escribe:

```python
import flet as ft
```

* Paso 2️⃣: Lista de proyectos

    Agrega tus proyectos de CreatiCode con nombre, descripción e imagen (opcional):

```python
PROJECTS = [
    {
        "nombre": "Dron Ciudad",
        "url": "https://play.creaticode.com/projects/690b5b1aecfdfbf9b9fe3d09/embed?version=1&autoPlay=true&btnOptions=none",
        "desc": "Recorre la ciudad 3D y elige el tipo de red ideal para cada zona.",
        "img": "https://i.ibb.co/Zcv6Hx8/dron-preview.gif",
    },
    {
        "nombre": "CPU — Museo 3D",
        "url": "https://play.creaticode.com/projects/69021536f80e3a9dbb616895/embed?version=1&autoPlay=true&btnOptions=none",
        "desc": "Explora las partes internas del CPU en un museo virtual.",
        "img": "https://i.ibb.co/3dgJvQj/placeholder-16x9.png",
    },
]
```

Cada proyecto es un diccionario con:

"nombre": título que aparecerá.

"url": enlace en modo player.

"desc": breve explicación.

"img": imagen o GIF opcional (si no tienes, deja el placeholder).

* Paso 3️⃣: Función para crear tarjetas

    Esto genera la presentación de cada proyecto.

```python
PLACEHOLDER = "https://i.ibb.co/3dgJvQj/placeholder-16x9.png"

def make_card(p, page):
    title = ft.Text(p["nombre"], size=18, weight=ft.FontWeight.BOLD, color=ft.Colors.CYAN_400)
    desc  = ft.Text(p["desc"], size=13, color=ft.Colors.WHITE70)
    img   = p.get("img") or PLACEHOLDER

    preview = ft.Image(src=img, height=180, fit=ft.ImageFit.COVER, border_radius=12)

    btn_open = ft.ElevatedButton(
        text="▶️ Abrir simulador",
        bgcolor=ft.Colors.CYAN_500,
        color="black",
        on_click=lambda e: page.launch_url(p["url"]),
    )

    return ft.Container(
        content=ft.Column([preview, title, desc, btn_open], spacing=10),
        bgcolor=ft.Colors.with_opacity(0.08, ft.Colors.WHITE),
        border_radius=16,
        padding=12,
    )
```

* Paso 4️⃣: Diseñar la página principal

    Ahora crea la función main que mostrará la galería:

```python
def main(page: ft.Page):
    page.title = "Galería de proyectos CreatiCode"
    page.theme_mode = ft.ThemeMode.DARK
    page.bgcolor = "#0d0d0d"
    page.padding = 20

    header = ft.Text("🚀 Galería de simuladores", size=26, weight=ft.FontWeight.BOLD, color=ft.Colors.CYAN_300)

    grid = ft.ResponsiveRow(run_spacing=16, spacing=16)

    for p in PROJECTS:
        grid.controls.append(
            ft.Container(content=make_card(p, page), col={"xs": 12, "sm": 6, "md": 4, "lg": 3})
        )

    footer = ft.Text("Desarrollado con Flet + CreatiCode", color=ft.Colors.WHITE54, size=12)

    page.add(header, grid, footer)
```

* Paso 5️⃣: Ejecutar la app

    Al final del archivo agrega:

```python
ft.app(target=main, view=ft.AppView.WEB_BROWSER)
```
